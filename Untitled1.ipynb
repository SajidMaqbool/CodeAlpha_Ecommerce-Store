```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, accuracy_score, confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Dataset Load Karein
df = pd.read_csv('eeg_dataset.csv')

# 2. Categorical variables ko numeric (One-Hot Encoding) karein
X = df[['age', 'gender', 'time_of_day', 'eeg_mean', 'eeg_std', 'eeg_max', 'eeg_min', 'valence', 'arousal']]
X_encoded = pd.get_dummies(X, columns=['gender', 'time_of_day'], drop_first=True)
y = df['stress']

# 3. Data Split (80% Train, 20% Test)
X_train, X_test, y_train, y_test = train_test_split(X_encoded, y, test_size=0.2, random_state=42)

# 4. Model Train Karein
model = RandomForestClassifier(n_estimators=100, random_state=42)
model.fit(X_train, y_train)

# 5. Predict & Evaluate
y_pred = model.predict(X_test)
print(f"Model Accuracy: {accuracy_score(y_test, y_pred) * 100:.2f}%\n")
print("Classification Report:\n", classification_report(y_test, y_pred))

# 6. Feature Importance Plot Karein
importance = pd.Series(model.feature_importances_, index=X_encoded.columns)
importance.nlargest(10).plot(kind='barh', title='Top Features for Stress Prediction')
plt.xlabel('Importance Score')
plt.show()
```

```python
import joblib

# Trained model aur feature list save karein
joblib.dump(model, 'eeg_stress_model.pkl')
joblib.dump(X_encoded.columns.tolist(), 'model_columns.pkl')
print("Model Successfully Saved!")
```

```python
import pandas as pd
import joblib

# Saved model load karein
loaded_model = joblib.load('eeg_stress_model.pkl')
model_cols = joblib.load('model_columns.pkl')

def predict_stress(age, gender, time_of_day, eeg_mean, eeg_std, eeg_max, eeg_min, valence, arousal):
    # Input DataFrame banayein
    input_data = pd.DataFrame([{
        'age': age,
        'gender': gender,
        'time_of_day': time_of_day,
        'eeg_mean': eeg_mean,
        'eeg_std': eeg_std,
        'eeg_max': eeg_max,
        'eeg_min': eeg_min,
        'valence': valence,
        'arousal': arousal
    }])

    # Preprocessing
    input_encoded = pd.get_dummies(input_data)
    input_encoded = input_encoded.reindex(columns=model_cols, fill_value=0)

    # Prediction
    prediction = loaded_model.predict(input_encoded)[0]
    stress_labels = {0: 'Low Stress', 1: 'Medium Stress', 2: 'High Stress'}

    return stress_labels[prediction]

# Sample Test
result = predict_stress(age=25, gender='M', time_of_day='afternoon',
                        eeg_mean=0.1, eeg_std=19.5, eeg_max=75.0,
                        eeg_min=-70.0, valence=2, arousal=1)
print("Predicted Mental State:", result)
```

```python
%%writefile app.py
import streamlit as st
import pandas as pd
import numpy as np
import joblib

# Page configuration
st.set_page_config(page_title="EEG Mental Health Assessment", page_icon="🧠", layout="centered")

st.title("🧠 EEG-Based Mental Health Assessment")
st.write("### Live Stress Prediction & Analysis System")
st.markdown("---")

# Load model and columns
@st.cache_resource
def load_model():
    model = joblib.load('eeg_stress_model.pkl')
    model_cols = joblib.load('model_columns.pkl')
    return model, model_cols

try:
    model, model_cols = load_model()
    st.sidebar.success("✅ ML Model Loaded Successfully")
except Exception as e:
    st.error("Model files not found. Please run the model saving script first.")

# Sidebar Inputs
st.sidebar.header("📋 Patient Demographics & EEG Inputs")

age = st.sidebar.slider("Age", 18, 60, 26)
gender = st.sidebar.selectbox("Gender", ["M", "F"])
time_of_day = st.sidebar.selectbox("Time of Day", ["morning", "afternoon", "evening"])

st.sidebar.subheader("⚡ Brainwave Statistics (EEG)")
eeg_mean = st.sidebar.number_input("EEG Mean Amplitude", value=0.08, format="%.4f")
eeg_std = st.sidebar.number_input("EEG Std Deviation", value=19.90, format="%.4f")
eeg_max = st.sidebar.number_input("EEG Max Peak", value=74.27, format="%.4f")
eeg_min = st.sidebar.number_input("EEG Min Peak", value=-77.30, format="%.4f")

st.sidebar.subheader("🎭 Emotional Metrics")
valence = st.sidebar.slider("Valence (0: Sad/Negative, 2: Happy/Positive)", 0, 2, 1)
arousal = st.sidebar.slider("Arousal (0: Calm, 2: Highly Excited)", 0, 2, 1)

# Display Selected Values in UI
col1, col2 = st.columns(2)
with col1:
    st.info(f"**Patient Age:** {age}\n\n**Gender:** {gender}\n\n**Session:** {time_of_day.capitalize()}")
with col2:
    st.info(f"**Valence Score:** {valence}\n\n**Arousal Score:** {arousal}\n\n**EEG Peak Range:** {round(eeg_max - eeg_min, 2)} µV")

# Predict Button
if st.button("🔍 Assess Stress Level", use_container_width=True):
    # Prepare dataframe for prediction
    input_df = pd.DataFrame([{
        'age': age,
        'gender': gender,
        'time_of_day': time_of_day,
        'eeg_mean': eeg_mean,
        'eeg_std': eeg_std,
        'eeg_max': eeg_max,
        'eeg_min': eeg_min,
        'valence': valence,
        'arousal': arousal
    }])

    # Preprocess matching training time
    encoded_df = pd.get_dummies(input_df)
    encoded_df = encoded_df.reindex(columns=model_cols, fill_value=0)

    # Predict
    prediction = model.predict(encoded_df)[0]
    probabilities = model.predict_proba(encoded_df)[0]

    st.markdown("---")
    st.subheader("📊 Diagnostic Assessment Result:")

    if prediction == 0:
        st.success("🟢 **LOW STRESS LEVEL (Normal / Relaxed State)**")
    elif prediction == 1:
        st.warning("🟡 **MODERATE STRESS LEVEL (Mild Anxiety / Fatigue)**")
    else:
        st.error("🔴 **HIGH STRESS LEVEL (Burnout / High Tension)**")

    # Show confidence metrics
    st.write("**Prediction Confidence:**")
    st.progress(float(np.max(probabilities)))
    st.caption(f"Confidence Score: {round(float(np.max(probabilities)) * 100, 2)}%")
```
